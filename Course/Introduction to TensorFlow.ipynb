```python
import tensorflow as tf

const_scalar = tf.constant(42)
const_vector = tf.constant([1.0, 2.0, 3.0])
const_matrix = tf.constant([[1.0, 2.0],
                             [3.0, 4.0]])
const_tensor = tf.constant([[[1, 2], [3, 4]],
                            [[5, 6], [7, 8]]])

var_scalar = tf.Variable(42)
var_vector = tf.Variable([1.0, 2.0, 3.0])
var_matrix = tf.Variable([[1.0, 2.0],
                           [3.0, 4.0]])
var_tensor = tf.Variable([[[1, 2], [3, 4]],
                          [[5, 6], [7, 8]]])
```

```python
print(f"Constant Scaler Shape : {const_scalar.shape} | Rank : {tf.rank(const_scalar).numpy()}")
print(f"Constant Vector Shape : {const_vector.shape} | Rank : {tf.rank(const_vector).numpy()}")
print(f"Constant Matrix Shape : {const_matrix.shape} | Rank : {tf.rank(const_matrix).numpy()}")
print(f"Constant Tensor Shapes : {const_tensor.shape} | Rank : {tf.rank(const_tensor).numpy()}")

vector_val = const_vector[2].numpy()
print(f"Element 3.0 from the Vector : {vector_val}")

matrix_val = const_matrix[1, 0].numpy()
print(f"Element 3.0 from the Matrix : {matrix_val}")

second_row = var_matrix[1, :].numpy()
print(f"Entire Second Row from the Matrix : {second_row}")

tensor_val_7 = const_tensor[1, 1, 0].numpy()
print(f"Element 7 from the Tensor : {tensor_val_7}")

first_matrix_block = var_tensor[0, :, :].numpy()
print(f"First 2 Dimensional block from the Tensor : \n {first_matrix_block}")
```

```python
const_vector = tf.constant([1.0, 2.0, 3.0])
const_matrix = tf.constant([[1.0, 2.0],
                             [3.0, 4.0]])

add_result = tf.add(const_vector, const_vector)
print("Addition : ", add_result.numpy())

sub_result = tf.subtract(const_vector, const_vector)
print("Subtraction : ", sub_result.numpy())

mul_result = tf.multiply(const_vector, const_vector)
print("Multiplication : ", mul_result.numpy())

div_result = tf.divide(const_vector, const_vector)
print("Division : ", div_result.numpy())

matmul_result = tf.matmul(const_matrix, const_matrix)

print("Original Matrix : \n", const_matrix.numpy())
print("\n Matrix Multiplication Result : \n", matmul_result.numpy())
```

```python
const_scalar = tf.constant(42)
const_vector = tf.constant([1.0, 2.0, 3.0])
const_matrix = tf.constant([[1.0, 2.0],
                             [3.0, 4.0]])
const_tensor = tf.constant([[[1, 2], [3, 4]],
                            [[5, 6], [7, 8]]])

flattened_matrix = tf.reshape(const_matrix, shape=[4])
print("Reshaped Matrix to 1D Vector :", flattened_matrix.numpy())
matrix_view = tf.reshape(var_tensor, shape=[2, 4])
print("Reshaped 3D Tensor to 2 x 4 Matrix : \n", matrix_view.numpy())

expanded_front = tf.expand_dims(const_vector, axis=0)
print(f"Expanded at Axis 0 Shape : {expanded_front.shape} | Data : {expanded_front.numpy()}")
expanded_back = tf.expand_dims(var_vector, axis=-1)
print(f"Expanded at Axis -1 Shape : {expanded_back.shape} | \n Data : \n {expanded_back.numpy()}")

squeezed_vector = tf.squeeze(expanded_front)
print(f"Before Squeeze Shape : {expanded_front.shape} | After Squeeze Shape : {squeezed_vector.shape}")

transposed_matrix = tf.transpose(const_matrix)
print("Transposed Matrix : \n", transposed_matrix.numpy())

permuted_tensor = tf.transpose(const_tensor, perm=[1, 2, 0])
print(f"Original 3D Tensor Shape : {const_tensor.shape} | Permuted Shape : {permuted_tensor.shape}")
 
stacked_vectors = tf.stack([const_vector, const_vector], axis=0)
print(f"Stacked Vectors Shape : {stacked_vectors.shape} \n Data : \n {stacked_vectors.numpy()}")

hstack_matrix = tf.experimental.numpy.hstack([const_matrix, const_matrix])
print("Horizontal Stack : \n ", hstack_matrix)

vstack_matrix = tf.experimental.numpy.vstack([const_matrix, const_matrix])
print("Vertical Stack : \n ", vstack_matrix)
vstack_vector = tf.experimental.numpy.vstack([const_vector, const_vector])
print("Vertical Stack on Vectors : \n ", vstack_vector)
```

```python
grades = tf.constant([
    [85.0, 92.0, 78.0], 
    [55.0, 60.0, 45.0],  
    [98.0, 40.0, 90.0],  
    [72.0, 88.0, 81.0]   
])

outlier_mask = (grades >= 90.0) | (grades < 60.0)
outlier_grades = tf.boolean_mask(grades, outlier_mask)

print("Grades Mask | Excellent or Failing : \n", outlier_mask.numpy())
print("Extracted Outlier Grades : ", outlier_grades.numpy())

inventory = tf.constant([
    [101.0,  5.0, 12.50],
    [102.0, 20.0, 45.00],
    [103.0,  0.0,  8.00],
    [104.0, 12.0, 99.00]
])

prices = inventory[:, 2]      
stock_levels = inventory[:, 1] 

stock_and_even_mask = (stock_levels > 0.0) & (prices % 1.0 == 0.0)
matching_prices = tf.boolean_mask(prices, stock_and_even_mask)

print("Inventory Price Match : ", matching_prices.numpy())

temperatures = tf.constant([22.5, -4.0, 18.0, -1.5, 31.2])

freezing_mask = temperatures < 0.0
not_freezing_mask = ~freezing_mask
warm_days = tf.boolean_mask(temperatures, not_freezing_mask)

print("Original Temperatures : ", temperatures.numpy())
print("Not Freezing Temperature : ", warm_days.numpy())

failing_coordinates = tf.where(grades < 60.0)
print("Coordinates of Failing Grades : \n", failing_coordinates.numpy())
```

```python
sensor_readings = tf.constant([1.5, -2.0, 4.8, 3.2, -0.5, 2.0])

total_sum = tf.reduce_sum(sensor_readings)
average   = tf.reduce_mean(sensor_readings)

variance  = tf.math.reduce_variance(sensor_readings)
std_dev   = tf.math.reduce_std(sensor_readings)
product   = tf.reduce_prod(sensor_readings)

print(f"Original Readings : {sensor_readings.numpy()}")
print(f"Sum : {total_sum.numpy():.2f} | Mean : {average.numpy():.2f}")
print(f"Variance : {variance.numpy():.2f} | Standard Deviation : {std_dev.numpy():.2f}")
print(f"Product Value : {product.numpy():.2f}")

predictions = tf.constant([
    [0.1, 0.8, 0.1],  
    [0.7, 0.2, 0.1], 
    [0.0, 0.3, 0.7],  
    [0.2, 0.6, 0.2]   
])

predicted_classes = tf.argmax(predictions, axis=1)
lowest_conf_index = tf.argmin(predictions, axis=1)

print("Predicted Class Indices : ", predicted_classes.numpy())
print("Lowest Confidence Indices : ", lowest_conf_index.numpy())

image_features = tf.constant([
    [[10, 20, 30], [40, 50, 60], [70, 80, 90]],  
    [[ 5, 15, 25], [35, 45, 55], [65, 75, 85]]   
])

spatial_means = tf.reduce_mean(tf.cast(image_features, tf.float32), axis=[1, 2])
print("Mean Values Calculated per Batch : ", spatial_means.numpy())

reduced_flat = tf.reduce_max(image_features, axis=0)
reduced_keep = tf.reduce_max(image_features, axis=0, keepdims=True)

print(f"Shape Without Dimensions : {reduced_flat.shape}") 
print(f"Shape With Dimensions : {reduced_keep.shape}") 

bool_tensor = tf.constant([
    [True,  True],
    [False, True]
])

print("Is Every Value True : ", tf.reduce_all(bool_tensor).numpy())
print("Is Any Value True :", tf.reduce_any(bool_tensor).numpy())
print("Column Check : ", tf.reduce_all(bool_tensor, axis=0).numpy())
```

```python
device = "/GPU:0" if tf.config.list_physical_devices('GPU') else "/CPU:0"
print(f"Targeted Device for Tensor Operations : {device}\n")

with tf.device(device):
    
    var_vector = tf.Variable([1.0, 2.0, 3.0])
    print("Initial Vector : ", var_vector.numpy())


    var_vector.scatter_update(
        sparse_delta=tf.IndexedSlices(
            values=[99.0],  
            indices=[1]     
        )
    )
    print("Updated Vector : ", var_vector.numpy())

    var_matrix = tf.Variable([[1.0, 2.0], 
                               [3.0, 4.0]])
    print("Initial Matrix : \n", var_matrix.numpy())

    var_matrix.scatter_nd_update(
        indices=[[1, 0]],  
        updates=[50.0]     
    )
    print("Updated Matrix : \n", var_matrix.numpy())

    var_matrix.scatter_update(
        sparse_delta=tf.IndexedSlices(
            values=[[7.0, 8.0]],
            indices=[0]         
        )
    )
    print("Matrix After Row Replacement : \n", var_matrix.numpy())
```
